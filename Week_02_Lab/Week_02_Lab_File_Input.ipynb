## Part 1: What is an even number?

An **even number** can be divided by 2 without leaving a remainder.

Examples:

- 2 is even.
- 8 is even.
- 14 is even.
- 7 is not even because dividing it by 2 leaves a remainder.

Python uses the `%` symbol to find the remainder.

```python
number % 2
```

When the remainder is `0`, the number is even:

```python
number % 2 == 0
```

The symbol `==` asks: **“Are these two values equal?”**


# Part 2: Learn the for loop before using the Excel file

## What is a for loop?

A **for loop** repeats instructions.

Imagine that you have a row of number cards:

```text
4, 7, 10, 13
```

A for loop picks up **one card at a time**:

1. Pick up 4.
2. Do the instruction.
3. Pick up 7.
4. Do the instruction.
5. Continue until there are no cards left.

The basic pattern is:

```python
for item in collection:
    instruction
```

- `for` tells Python to start a loop.
- `item` is the current value.
- `in` means “take values from”.
- `collection` is the group of values.
- The indented line is repeated.


````python
# A very simple for loop

numbers = [4, 7, 10, 13]

for number in numbers:
    print(number)
````

### Explanation of the previous cell

```python
numbers = [4, 7, 10, 13]
```

This creates a **list** named `numbers`.

A list stores several values inside square brackets `[ ]`.

```python
for number in numbers:
```

This means:

> Take one value at a time from the `numbers` list and temporarily call it `number`.

```python
print(number)
```

This prints the current number.

The output should be:

```text
4
7
10
13
```

Notice that `print(number)` is indented. The indentation tells Python that this instruction belongs to the loop.

## Add an if statement inside a for loop

An `if` statement lets Python make a decision.

```python
if condition:
    instruction
```

In the next example:

1. The for loop checks one number at a time.
2. The if statement checks whether the number is even.
3. Python prints only the even numbers.


````python
practice_numbers = [3, 4, 7, 8, 11, 12]

for number in practice_numbers:
    if number % 2 == 0:
        print(number, "is even")
````

### Line-by-line explanation

```python
practice_numbers = [3, 4, 7, 8, 11, 12]
```

Creates a list of numbers.

```python
for number in practice_numbers:
```

Takes one number at a time from the list.

```python
if number % 2 == 0:
```

Checks whether dividing the number by 2 leaves a remainder of 0.

```python
print(number, "is even")
```

Prints the number only when the condition is true.

The expected output is:

```text
4 is even
8 is even
12 is even
```

# Part 4: Read the Excel file with Pandas

## What is Pandas?

**Pandas** is a Python library used to work with data arranged in rows and columns, like an Excel spreadsheet.

A **library** is a collection of useful code written by other programmers.

```python
import pandas as pd
```

- `import pandas` loads the Pandas library.
- `as pd` gives Pandas a short nickname: `pd`.
- Instead of writing `pandas.read_excel(...)`, we can write `pd.read_excel(...)`.

The nickname does not change Pandas. It only makes the code shorter.


````python
# Step 2A: Import Pandas

import pandas as pd

print("Pandas is ready.")
````

## Find the uploaded filename

The variable `uploaded` stores information in a dictionary-like structure.

We need to get the name of the uploaded file.

```python
iter(uploaded)
```

- `iter` means **iterator**.
- An iterator lets Python move through items one at a time.
- Here, it lets Python move through the uploaded filenames.

```python
next(iter(uploaded))
```

- `next()` asks for the first available item from the iterator.
- Because we upload one file, this gives us that file's name.

```python
file_name = next(iter(uploaded))
```

This stores the uploaded filename in a variable called `file_name`.

For example, the value might be:

```text
Even_Odd_Numbers.xlsx
```


````python
# Step 2B: Get the name of the uploaded file

file_name = next(iter(uploaded))

print("The uploaded filename is:")
print(file_name)
````

## Read the Excel file

```python
data = pd.read_excel(file_name)
```

Meaning:

- `pd` refers to the Pandas library.
- `read_excel` is a Pandas function that opens an Excel file.
- `file_name` tells the function which file to open.
- The round brackets contain the information given to the function.
- `data` stores the spreadsheet in Python.

Pandas stores the spreadsheet as a **DataFrame**.

A DataFrame is like a table with:

- rows,
- columns,
- column headings,
- and data values.

```python
print("Numbers in the file:")
```

Prints a heading.

```python
display(data)
```

Displays the DataFrame as a clear table in Google Colab.


````python
# Step 2C: Read and display the Excel file

data = pd.read_excel(file_name)

print("Numbers in the file:")
display(data)
````

# Part 5: Understand the spreadsheet column

The Excel file has a column called:

```text
Number
```

We can select that column using:

```python
data["Number"]
```

- `data` is the whole table.
- `["Number"]` selects only the column named `Number`.

The next cell prints the values from that column.


````python
# Look at the Number column

print(data["Number"])
````

# Part 6: Find the even numbers

First, we create an empty list:

```python
even_numbers = []
```

The empty square brackets mean that the list has no items yet.

Then we use:

```python
for number in data["Number"]:
```

This takes one value at a time from the Excel column named `Number`.

Next:

```python
if number % 2 == 0:
```

This checks whether the current number is even.

Finally:

```python
even_numbers.append(number)
```

- `append` means **add an item to the end of a list**.
- If the current number is even, Python adds it to `even_numbers`.


````python
# Step 3: Find the even numbers

even_numbers = []

for number in data["Number"]:
    if number % 2 == 0:
        even_numbers.append(number)

print("Even numbers are:")
print(even_numbers)
````

## Show each even number with a message

The list `even_numbers` now contains only even numbers.

```python
for number in even_numbers:
```

takes one even number at a time.

```python
print(number, "is an even number")
```

prints a friendly sentence for each number.


````python
# Step 4: Show each even number

for number in even_numbers:
    print(number, "is an even number")
````

# Part 7: Find the odd numbers

An odd number leaves a remainder when divided by 2.

```python
number % 2 != 0
```

- `!=` means **not equal to**.
- Therefore, this condition means:
  “The remainder is not equal to zero.”

The process is similar to finding even numbers:

1. Create an empty list.
2. Check each number.
3. Add odd numbers to the list.


````python
# Step 5: Find the odd numbers

odd_numbers = []

for number in data["Number"]:
    if number % 2 != 0:
        odd_numbers.append(number)

print("Odd numbers are:")
print(odd_numbers)
````

# Part 8: Count the results

Python's `len()` function counts the number of items in a list.

```python
len(even_numbers)
```

counts the even numbers.

```python
len(odd_numbers)
```

counts the odd numbers.


````python
# Step 6: Count even and odd numbers

print("Number of even numbers:", len(even_numbers))
print("Number of odd numbers:", len(odd_numbers))
````

# Part 9: Create a results table

```python
pd.DataFrame(...)
```

creates a new Pandas table.

The next cell creates two columns:

- `Even Numbers`
- `Odd Numbers`

Because the two lists may have different lengths, we use `pd.Series()` to allow empty spaces where needed.


````python
# Step 7: Create a result table

result_table = pd.DataFrame({
    "Even Numbers": pd.Series(even_numbers),
    "Odd Numbers": pd.Series(odd_numbers)
})

display(result_table)
````

# Part 10: Save the even numbers into a new Excel file

```python
even_data = pd.DataFrame({"Even Number": even_numbers})
```

Creates a new table containing the even numbers.

```python
even_data.to_excel("Even_Numbers_Result.xlsx", index=False)
```

- `to_excel` saves the table as an Excel file.
- `"Even_Numbers_Result.xlsx"` is the new filename.
- `index=False` stops Pandas from adding an extra row-number column.

```python
files.download("Even_Numbers_Result.xlsx")
```

Downloads the new file from Google Colab to your computer.


````python
# Step 8: Save and download the even-number results

even_data = pd.DataFrame({"Even Number": even_numbers})

even_data.to_excel(
    "Even_Numbers_Result.xlsx",
    index=False
)

files.download("Even_Numbers_Result.xlsx")
````
