# Step 1: What is a Database?

A **database** is a place where information is stored in an organised way.

For example, a school might store:

| Student ID | Name | Subject | Mark |
|---|---|---|---|
| 1 | Emma | Python | 85 |
| 2 | Noah | Database | 78 |
| 3 | Mia | Python | 92 |

A database helps us:
- save information
- search information
- change information
- remove information

These actions are called **CRUD**.

# Step 2: What is SQLite?

**SQLite** is a simple relational database.

Relational databases usually keep information in **tables**.

A table has:
- **Rows** → each row is one record
- **Columns** → each column stores one type of information

Example:

| id | name | subject | mark |
|---|---|---|---|
| 1 | Emma | Python | 85 |

In this notebook, Python will create an SQLite database file for us.

# Step 3: Import the Python Libraries

We need two libraries.

### `sqlite3`
This lets Python talk to SQLite.

### `pandas`
This helps us show database results as neat tables.

````python
import sqlite3
import pandas as pd

print("Great! The libraries are ready.")
````

# Step 4: Connect Python to the Database

We now create or open a database called:

`school_database.db`

If the file does not exist, Python creates it.

If it already exists, Python opens it.

````python
connection = sqlite3.connect("school_database.db")

print("Connected to school_database.db")
````

# Step 5: Create a Cursor

A **cursor** is like a messenger.

It takes our SQL instructions from Python and gives them to the database.

````python
cursor = connection.cursor()

print("Cursor is ready.")
````

# Step 6: Create a Students Table

We will make a table called `students`.

It will contain:

- `student_id` → unique number for each student
- `name` → student's name
- `subject` → subject name
- `mark` → student's mark

````python
cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    student_id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    subject TEXT NOT NULL,
    mark INTEGER
)
""")

connection.commit()

print("The students table is ready.")
````

# Step 7: Look at the Table

At first, the table may be empty.

We can use `SELECT * FROM students` to read every row.

````python
students_df = pd.read_sql_query(
    "SELECT * FROM students",
    connection
)

students_df
````

# CRUD Part 1: CREATE

## What does CREATE mean?

**CREATE** means adding new information to the database.

In SQL, we often use:

```sql
INSERT INTO
```

Imagine we want to add:

- Name: Emma
- Subject: Python
- Mark: 85

````python
cursor.execute(
    "INSERT INTO students (name, subject, mark) VALUES (?, ?, ?)",
    ("Emma", "Python", 85)
)

connection.commit()

print("Emma was added to the database.")
````

# Add More Students

We can add several students at once.

This is useful when we have many records.

````python
more_students = [
    ("Noah", "Database", 78),
    ("Mia", "Python", 92),
    ("Oliver", "Cyber Security", 74),
    ("Ava", "Database", 88)
]

cursor.executemany(
    "INSERT INTO students (name, subject, mark) VALUES (?, ?, ?)",
    more_students
)

connection.commit()

print("More students were added.")
````

# Check the New Data

Now let's read the table again.

````python
pd.read_sql_query(
    "SELECT * FROM students",
    connection
)
````

# CRUD Part 2: READ

## What does READ mean?

**READ** means looking at information stored in the database.

In SQL we use:

```sql
SELECT
```

Let's display all students.

````python
all_students = pd.read_sql_query(
    "SELECT * FROM students",
    connection
)

all_students
````

# Read Only Some Columns

Maybe we do not need every column.

We can ask for only:
- name
- subject
- mark

````python
pd.read_sql_query(
    "SELECT name, subject, mark FROM students",
    connection
)
````

# Read Students With High Marks

We can use `WHERE` to choose only some rows.

Let's find students with marks of **80 or higher**.

````python
pd.read_sql_query(
    "SELECT * FROM students WHERE mark >= 80",
    connection
)
````

# Read Students From One Subject

Let's find only students studying Python.

````python
pd.read_sql_query(
    "SELECT * FROM students WHERE subject = 'Python'",
    connection
)
````

# CRUD Part 3: UPDATE

## What does UPDATE mean?

**UPDATE** means changing information that already exists.

For example:

Maybe Noah's mark was entered as 78, but the teacher later changes it to 82.

We use:

```sql
UPDATE
```

````python
cursor.execute(
    "UPDATE students SET mark = ? WHERE name = ?",
    (82, "Noah")
)

connection.commit()

print("Noah's mark was updated.")
````

# Check Noah's New Mark

````python
pd.read_sql_query(
    "SELECT * FROM students WHERE name = 'Noah'",
    connection
)
````

# Another UPDATE Example

Let's change Oliver's subject from `Cyber Security` to `Python`.

````python
cursor.execute(
    "UPDATE students SET subject = ? WHERE name = ?",
    ("Python", "Oliver")
)

connection.commit()

print("Oliver's subject was updated.")
````

````python
pd.read_sql_query(
    "SELECT * FROM students WHERE name = 'Oliver'",
    connection
)
````

# CRUD Part 4: DELETE

## What does DELETE mean?

**DELETE** means removing a record from the database.

For example, if a student record was added by mistake, we can remove it.

In SQL we use:

```sql
DELETE FROM
```

````python
cursor.execute(
    "DELETE FROM students WHERE name = ?",
    ("Ava",)
)

connection.commit()

print("Ava's record was deleted.")
````

# Check the Table After DELETE

````python
pd.read_sql_query(
    "SELECT * FROM students",
    connection
)
````

# CRUD Summary

| CRUD Letter | Meaning | SQL Command | Example |
|---|---|---|---|
| C | Create | `INSERT` | Add Emma |
| R | Read | `SELECT` | Show all students |
| U | Update | `UPDATE` | Change Noah's mark |
| D | Delete | `DELETE` | Remove Ava |

This is the basic CRUD cycle used in many real applications.

# Step 8: Count the Students

We can also use SQL to count records.

````python
pd.read_sql_query(
    "SELECT COUNT(*) AS total_students FROM students",
    connection
)
````

# Step 9: Find the Average Mark

We can ask the database to calculate the average mark.

````python
pd.read_sql_query(
    "SELECT ROUND(AVG(mark), 2) AS average_mark FROM students",
    connection
)
````

# Step 10: Average Mark for Each Subject

This shows how database data can also be analysed.

````python
pd.read_sql_query("""
SELECT subject, ROUND(AVG(mark), 2) AS average_mark
FROM students
GROUP BY subject
ORDER BY average_mark DESC
""", connection)
````

# Step 11: Sort Students by Mark

Let's show the highest mark first.

````python
pd.read_sql_query(
    "SELECT * FROM students ORDER BY mark DESC",
    connection
)
````

# Step 12: Using Python Variables With SQL

We can store values in Python variables and use them in our query.

This is safer than joining text together.

````python
minimum_mark = 80

pd.read_sql_query(
    "SELECT * FROM students WHERE mark >= ?",
    connection,
    params=(minimum_mark,)
)
````

# Step 13: Add One Student Using Variables

Let's create the student's details as normal Python variables.

````python
student_name = "Liam"
student_subject = "Python"
student_mark = 90

cursor.execute(
    "INSERT INTO students (name, subject, mark) VALUES (?, ?, ?)",
    (student_name, student_subject, student_mark)
)

connection.commit()

print("Liam was added.")
````

# Step 14: Display the Final Table

````python
final_table = pd.read_sql_query(
    "SELECT * FROM students",
    connection
)

final_table
````

# Step 15: Save Database Data to CSV

We can save the table as a CSV file.

CSV files can be opened using:
- Excel
- Google Sheets
- Python

````python
final_table.to_csv(
    "students_from_database.csv",
    index=False
)

print("students_from_database.csv was created.")
````

# Important Database Rule

After making changes such as:

- INSERT
- UPDATE
- DELETE

we use:

```python
connection.commit()
```

This saves the change.

Without `commit()`, the change may not be permanently saved.

When we are finished, we should close the database connection.
