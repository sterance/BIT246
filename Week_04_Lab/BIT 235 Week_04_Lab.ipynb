## Part 1: Start with a very simple class
A **class** is a design for creating objects.

A **Student** object can store:
- name
- mark

A **method** can calculate whether the student passed.


````python

class Student:
    # __init__ runs when we create a new Student object.
    def __init__(self, name, mark):
        # self.name stores the name inside this object.
        self.name = name
        self.mark = mark

    # result() is a method.
    # It uses the mark stored inside the object.
    def result(self):
        if self.mark >= 50:
            return "Pass"
        else:
            return "Fail"

# Create two Student objects.
s1 = Student("Ana", 78)
s2 = Student("Ben", 42)

print(s1.name, s1.result())
print(s2.name, s2.result())
````


## Part 4: Create a NumberRecord class
Each row in the Excel file contains one number.

We will create one object for each number.


````python

class NumberRecord:
    def __init__(self, number):
        self.number = number

    def is_even(self):
        return self.number % 2 == 0

    def number_type(self):
        if self.is_even():
            return "Even"
        else:
            return "Odd"

    def show(self):
        return f"{self.number} is {self.number_type()}"
````


## Part 5: Convert Excel rows into objects
The file data is in a table.

We will:
1. loop through the Number column,
2. create a `NumberRecord` object,
3. store the object in a list.


````python

number_objects = []

for number in numbers_df["Number"]:
    obj = NumberRecord(number)
    number_objects.append(obj)

print("Number of objects created:", len(number_objects))

# Show the first object.
print(number_objects[0].show())
````


## Part 6: Use methods to print all results
Each object has its own `show()` method.


````python

for obj in number_objects:
    print(obj.show())
````


## Part 7: Filter objects using a method
Now we will print only the even numbers.


````python

print("Even numbers:")

for obj in number_objects:
    if obj.is_even():
        print(obj.number)
````


## Part 9: AccidentReport class
Each row in the CSV file is one accident report.

The class will store:
- road
- weather
- speed
- helmet
- signal
- severity


````python

class AccidentReport:
    def __init__(self, row):
        # row is one dictionary-like row from the DataFrame.
        self.road = row["road"]
        self.weather = row["weather"]
        self.speed = row["speed"]
        self.helmet = row["helmet"]
        self.signal = row["signal"]
        self.severity = row["severity"]

    def show(self):
        return f"{self.road} | {self.weather} | {self.speed} km/h | {self.severity}"

    def is_high_risk(self):
        return self.severity == "High"

    def is_fast(self):
        return self.speed > 60
````


## Part 10: Convert CSV rows into AccidentReport objects
We will use `iterrows()` to read one row at a time from the DataFrame.


````python

accident_objects = []

for index, row in accident_df.iterrows():
    report = AccidentReport(row)
    accident_objects.append(report)

print("Number of accident objects:", len(accident_objects))
print("First accident object:")
print(accident_objects[0].show())
````


## Part 11: Use object methods to answer questions
Question: Which accident reports are high risk?


````python

print("High risk reports:")

for report in accident_objects:
    if report.is_high_risk():
        print(report.show())
````


## Part 12: Use another method
Question: Which accident reports have speed greater than 60?


````python

print("Fast accident reports:")

for report in accident_objects:
    if report.is_fast():
        print(report.show())
````
