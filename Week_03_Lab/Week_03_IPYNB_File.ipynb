## Part 2: Lists
A **list** stores many items **in order**.

Use a list when:
- order matters
- items may repeat
- you want to store many records together


```python
# The whole dataset is stored as a list of records
print(type(records))
print("First record:")
print(records[0])
```

```python
# A list can also store only one field from each record
roads_list = [record['road'] for record in records]
print("Road list:")
print(roads_list)
```

## Part 3: Dictionaries
A **dictionary** stores one record using **labels** called keys.

Use a dictionary when:
- each value has a name
- you want quick lookup by key
- you want one well-organised record


```python
# One accident report as a dictionary
one_report = records[0]
print(type(one_report))
print(one_report)

print("\nRoad:", one_report['road'])
print("Speed:", one_report['speed'])
```

## Part 4: Sets
A **set** stores **unique values only**.

Use a set when:
- you want duplicates removed
- you want a quick membership check
- order does not matter


```python
# Unique roads and unique weather types
unique_roads = {record['road'] for record in records}
unique_weather = {record['weather'] for record in records}

print("Unique roads:", unique_roads)
print("Unique weather values:", unique_weather)
```

## Part 5: When is each one best?
- **List**: best for many records in order.
- **Dictionary**: best for one labeled record.
- **Set**: best for unique values.


```python
print("LIST is best for the full dataset.")
print("DICTIONARY is best for one accident record.")
print("SET is best for unique roads or weather values.")
```

## Part 6: Functions
A **function** is a reusable mini-job.

We will create small functions to work with our accident data.


```python
# Function 1: classify speed into a message
def speed_message(speed):
    if speed > 60:
        return "High speed"
    elif speed > 45:
        return "Medium speed"
    else:
        return "Lower speed"

print(speed_message(72))
print(speed_message(40))
```

```python
# Function 2: summarise one accident report
def report_summary(report):
    return f"{report['road']} | {report['weather']} | {report['speed']} km/h | {report['severity']} severity"

print(report_summary(records[0]))
```

## Part 7: File Functions
We will save the dataset to a CSV file and then read it back.

CSV means **comma-separated values**. It is a simple file format for tables.


```python
# Turn the list of dictionaries into a DataFrame
accident_df = pd.DataFrame(records)
accident_df.head()
```

```python
# Read the file back into Python
loaded_df = pd.read_csv('road_accidents.csv')
print('File loaded successfully.')
loaded_df.head()
```

## Part 8: Classes
A **class** is a plan for creating similar objects.

We will create a class called `AccidentReport`.


```python
class AccidentReport:
    def __init__(self, road, weather, speed, helmet, signal, severity):
        self.road = road
        self.weather = weather
        self.speed = speed
        self.helmet = helmet
        self.signal = signal
        self.severity = severity

    def show(self):
        return f"Road: {self.road}, Weather: {self.weather}, Speed: {self.speed}, Severity: {self.severity}"

    def is_high_risk(self):
        return self.severity == 'High'
```

```python
# Create one object from the first record
first = records[0]
report_object = AccidentReport(
    first['road'],
    first['weather'],
    first['speed'],
    first['helmet'],
    first['signal'],
    first['severity']
)

print(report_object.show())
print("High risk?", report_object.is_high_risk())
```

## Part 9: Decision Tree
A **decision tree** asks simple questions to make a prediction.

We will predict **severity** from:
- weather
- speed
- helmet
- signal


```python
# Prepare the data for a machine learning model
ml_df = loaded_df.copy()
X = pd.get_dummies(ml_df[['weather', 'speed', 'helmet', 'signal']])
y = ml_df['severity']

print('Input columns:')
print(list(X.columns))
print('\nTarget values:')
print(y.unique())
```

```python
# Train a simple decision tree
model = DecisionTreeClassifier(max_depth=3, random_state=42)
model.fit(X, y)
print('Decision tree trained successfully.')
```

```python
# Show the decision tree
plt.figure(figsize=(16, 8))
plot_tree(
    model,
    feature_names=X.columns,
    class_names=model.classes_,
    filled=True,
    rounded=True,
    fontsize=10
)
plt.title('Decision Tree for Accident Severity')
plt.show()
```

```python
# Predict one new accident case
new_case = pd.DataFrame([{'weather': 'Rainy', 'speed': 72, 'helmet': 'No', 'signal': 'Red'}])
new_case_encoded = pd.get_dummies(new_case)
new_case_encoded = new_case_encoded.reindex(columns=X.columns, fill_value=0)
prediction = model.predict(new_case_encoded)[0]
print('Predicted severity for the new case:', prediction)
```
